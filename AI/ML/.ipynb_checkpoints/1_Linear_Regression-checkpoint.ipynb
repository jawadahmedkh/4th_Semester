### `Percent of Peoples lies outside 1st Standard Deviation`

```python
# Calculating Q1, Q2 and IQR
Q1 = medical_df['charges'].quantile(0.25)
Q3 = medical_df['charges'].quantile(0.75)
IQR = Q3 - Q1

# Finding Lower and Upper Bound for finding the outliers.
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Finding Outliers
outliers = medical_df[(medical_df['charges'] < lower_bound) | (medical_df['charges'] > upper_bound)]

#Finding Percentage
percent_outside = (len(outliers) / len(medical_df)) * 100
print(f"Percentage of people outside the norm (by IQR): {percent_outside:.2f}%")

```

`OBSERVATIONS`

From above code I come to know that there are `10.39%` Peoples outside first norm.

# `5. Correlation`

> Correlation is a statistical measure that describes the strength and direction of a relationship between two variables. It represents `Strength` and `Direction`.
> 

`Formula of Correlation (Pearson’s Correlation Coefficient):`

$$
r = \frac{ \sum (x_i - \bar{x})(y_i - \bar{y}) }{ \sqrt{ \sum (x_i - \bar{x})^2 } \cdot \sqrt{ \sum (y_i - \bar{y})^2 } }
$$

Where:

* $r$ = Pearson correlation coefficient
* $x_i, y_i$ = Individual sample points
* $\bar{x}, \bar{y}$ = Means of x and y
* $\sum$ = Summation over all data points


`Meaning of Correlation:`

**Correlation** measures the **strength** and **direction** of a **linear relationship** between two variables.

| Value of $r$         | Meaning                                                    |
| -------------------- | ---------------------------------------------------------- |
| $r = 1$              | Perfect positive correlation (as x increases, y increases) |
| $r = -1$             | Perfect negative correlation (as x increases, y decreases) |
| $r = 0$              | No linear correlation                                      |
| $0.7 \leq r < 1$     | Strong positive correlation                                |
| $0.3 \leq r < 0.7$   | Moderate positive correlation                              |
| $0 < r < 0.3$        | Weak positive correlation                                  |
| $-0.3 < r < 0$       | Weak negative correlation                                  |
| $-0.7 < r \leq -0.3$ | Moderate negative correlation                              |
| $-1 < r \leq -0.7$   | Strong negative correlation                                |


`Shortcut Formula (for sample data):`

$$
r = \frac{ n\sum xy - \sum x \sum y }{ \sqrt{ [n\sum x^2 - (\sum x)^2][n\sum y^2 - (\sum y)^2] } }
$$

Where:

* $n$ = number of data points
* $\sum xy$ = sum of product of x and y

# `6. Modeling`

## `6.1. Linear Regression`

**Linear Regression** is a statistical method used to model the relationship between a **dependent variable** (target/output) and one or more **independent variables** (inputs/features). The goal is to find a **line of best fit** that can predict the output based on input values.


### 🔹 **Simple Linear Regression** (One Independent Variable)

It fits a straight line:

$$
y = mx + c
$$

* **y**: Dependent variable (what you’re trying to predict)
* **x**: Independent variable (the input feature)
* **m**: Slope of the line (shows how much y changes with x)
* **c**: Intercept (the value of y when x = 0)

### 🔹 **Multiple Linear Regression** (More than One Independent Variable)

$$
y = b_0 + b_1x_1 + b_2x_2 + \dots + b_nx_n
$$

* It models the relationship using a **hyperplane** instead of a line.
* Each $b_i$ is a coefficient showing how that feature influences $y$.

### 🔹 Use Cases:

* Predicting house prices based on features like size, location, etc.
* Estimating sales based on advertising budget.
* Forecasting trends in economics or finance.

### 🔹 Assumptions of Linear Regression:

1. Linearity between input and output.
2. Homoscedasticity (equal variance of errors).
3. Independence of observations.
4. Normal distribution of residuals.

1. Now to model it we have to tell computer that how can computer see the line has to go upward, lower, right or left?

For this we use `RMSE` i.e. `Lost/Cost Function`

---

### `6.1.1. RMSE (Root Mean Squared Error)`

1. First we measure the Distance of line from target by calculating `Residuals`.
2. What we actually do is we subtract every value of predicated values from target values then we can know how far we are from the targets also one more thing the answer value maybe in negative so we have to take care of it also, `that is the reason we take the square these values to make them positive` and now we taken an extra value by squaring, we have to remove it, `that is the reason we have taken a square root on these values`

`NOTE: This is Sensitive to Outliers`




```python
import numpy as np

def rmse(predicted,actual):
    return float(np.sqrt(np.mean(np.square(predicted-actual))))
```

```python
targets = non_smoker_df.charges
predicted = estimate_charges(ages,50,100)

rmse(predicted,targets)
```

`OBSERVATIONS`

From above Number we can say that we have difference of $8461 from actual charges on average.


Let's Modify the `try_parameters` Function.

```python
def try_parameters(m,c):
    ages = non_smoker_df.age
    target_charges = non_smoker_df.charges

    estimated_charges = estimate_charges(ages,m,c)

    plt.plot(ages, estimated_charges, 'r-', alpha=0.9)
    plt.scatter(ages, target_charges, s=8, alpha=0.8)
    plt.xlabel('Ages')
    plt.ylabel('Charges')
    plt.legend(['Estimated Charges','Actual Charges'])

    loss = rmse(estimated_charges,target_charges)
    print(f'RMSE: {loss:.2f}')

    
```

```python
try_parameters(300,-3000)
```

`OBSERVATIONS`

From above `RMSE Value: $4700.736524304107` I think it could be the best value I have got for the Charges.

## `3. Linear Regression in sklearn Using SGDRegressor Class`

In `LinearRegression()` sklearn uses `Ordinary Least Square Method` for optimization, for using `Stochastic Gradient Descent` We have to use `SGDRegressor()`.

`NOTE: The SGDRegressor is sensitivte to Feature Scaling, Do it first if you want best result.`

### `3.1. Feature Scaling`

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
inputs = scaler.fit_transform(non_smoker_df[['age']] ) 
```

### `3.2. Now Creating Model`

```python
target = non_smoker_df.charges

from sklearn.linear_model import SGDRegressor

model = SGDRegressor()
model.fit(inputs,target)
predictions = model.predict(inputs)

plt.plot(inputs, predictions, 'r-', alpha=0.9)
plt.scatter(inputs, target, s=8, alpha=0.8)
plt.xlabel('Ages')
plt.ylabel('Charges')
plt.legend(['Estimated Charges','Actual Charges'])



rmse(predictions, target)
```

**Let's Create Both Models in a single tab**

`Linear Regression`

```python
# Create Inputs and Targets

inputs, targets = non_smoker_df[['age']], non_smoker_df.charges

# Create and Train Model

model = LinearRegression().fit(inputs,targets)

# Generate Predictions

predictions = model.predict(inputs)

# Calculate Loss to Evaluate the model

loss = rmse(predictions,target)

print(f'Loss: {loss}')
```

```python
# Create and Scale the Data
scaler = StandardScaler()
inputs = scaler.fit_transform(non_smoker_df[['age']] ) 
target = non_smoker_df.charges

# Create and Train Model
model = SGDRegressor().fit(inputs,target)

# Generate Predictions
predictions = model.predict(inputs)


# Calculate Loss to Evaluate the model
loss = rmse(predictions,target)

print(f'Loss: {loss}')
```
